```python
import heapq
goal=(1,2,3,4,5,6,7,8,0)
def minkowski(state):
  dist=0
  p=0.5
  for i in range(9):
    if state[i]!=0:
      goal_pos=goal.index(state[i])
    row=abs(i//3-goal_pos//3)
    col=abs(i%3-goal_pos%3)
    cal=(row**p+col**p)**(1/p)
    dist+=cal
  return dist
```

```python
def get_neighbours(state):
  neighbours=[]
  idx=state.index(0)
  moves={-3,3,-1,1}
  for m in moves:
    new=idx+m
    if 0 <= new < 9:
      new_state=list(state)
      new_state[idx],new_state[new]=new_state[new],new_state[idx]

      neighbours.append(tuple(new_state))

  return neighbours
```

```python
import heapq
def astar(start):
  pq=[(minkowski(start),0,start)]
  visited=set()
  i=1
  while pq:
    f,g,state=heapq.heappop(pq)

    if(state)==goal:
      print("solved")
      print(state)
      print(i)
      return
    if state in visited:
      continue
    print(state)
    visited.add(state)

    for n in get_neighbours(state):
      heapq.heappush(pq,(g+1+minkowski(n),g+1,n))
    i=i+1
start=tuple(int(x) for x in input().split())
astar(start)
```
